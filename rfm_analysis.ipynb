```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("../data/processed/data_preprocessed.csv",encoding='latin-1')
data.head()
```

```python
df = data[
    [
        'Customer Id',
        'Order Id',
        'Sales',
        'order date (DateOrders)',
        'Department Name',
        'Order Item Discount',
        'Order Profit Per Order',
        'Order Item Discount Rate'
    ]
]
df
```

Creating Recency, Frequency and Monetary(RFM)

```python
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])
snapshot_date = (df['order date (DateOrders)'].max()+ pd.Timedelta(days=1))
rfm = (df.groupby('Customer Id').agg({'order date (DateOrders)': lambda x:(snapshot_date - x.max()).days,
                                      'Order Id': 'nunique',
                                      'Sales': 'sum'}).reset_index())
rfm.columns = ['Customer Id','Recency','Frequency','Monetary']
rfm
```

Average Order Value

```python
rfm['Avg_Order_Value']=rfm['Monetary']/rfm['Frequency']
```

Product Category Diversity

```python
category_diversity = (df.groupby("Customer Id")["Department Name"].nunique())
rfm['Unique_Product_Categories']=category_diversity
rfm.head()
```

Discount Behavior

```python
# Average discount rate per customer
avg_discount_rate = (df.groupby("Customer Id")["Order Item Discount Rate"].mean())
rfm['Avg_Discount_Rate']=round(avg_discount_rate,2)
rfm.head()
```

Total Profitability

```python

#Aggregate the additional features by customer ID
extra_features = (df.groupby("Customer Id").agg(
        Unique_Product_Categories=("Department Name", "nunique"),
        Avg_Discount_Rate=("Order Item Discount Rate", "mean"),
        Total_Profit=("Order Profit Per Order", "sum")).reset_index())

```

```python
#Merge using the actual customer ID
rfm = rfm.drop(columns=["Unique_Product_Categories","Avg_Discount_Rate","Total_Profit"],errors="ignore")
rfm = rfm.merge(extra_features,on="Customer Id",how="left",validate="one_to_one")
```

```python
#Check the result
print("RFM shape:", rfm.shape)
print("Unique customers:", rfm["Customer Id"].nunique())
print("\nMissing values in engineered features:")
print(rfm[["Unique_Product_Categories",
            "Avg_Discount_Rate",
            "Total_Profit"]].isna().sum())
print("\nDuplicate customer IDs:", rfm["Customer Id"].duplicated().sum())
rfm.head()
```

```python
numeric_columns = ["Recency","Frequency","Monetary","Avg_Order_Value","Unique_Product_Categories","Avg_Discount_Rate","Total_Profit"]

skewness = rfm[numeric_columns].skew().sort_values(ascending=False)
print(skewness)
```

```python
rfm.to_csv("rfm data/rfm.csv", index=False)
```
